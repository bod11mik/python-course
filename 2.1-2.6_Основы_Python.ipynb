## Часть 2. Операторы (2.3)

**Оператор** - символ, выполняющий операцию над значениями. Данные и операторы вместе образуют **выражение**.

| Оператор | Операция | Пример | Результат |
|---|---|---|---|
| `**` | возведение в степень | `2 ** 3` | `8` |
| `*` | умножение | `2 * 3` | `6` |
| `/` | деление | `6 / 3` | `2.0` |
| `//` | целочисленное деление | `6 // 4` | `1` |
| `%` | остаток от деления | `14 % 4` | `2` |
| `+` `-` | сложение и вычитание | `-4 + 4` | `0` |

**Три правила, которые надо запомнить:**

1. **Целое против float.** Если оба аргумента целые - результат целый. Если хотя бы один float - результат float.
2. **Исключение: `/` всегда даёт float.** Даже `4 / 2` даёт `2.0`, а не `2`.
3. **`//` всегда округляет вниз**, в сторону меньшего целого. Поэтому `-6 // 4` даёт `-2`, а не `-1` (настоящий результат `-1.5`, а меньшее целое здесь `-2`).

Деление на ноль (`/`, `//`, `%`) - ошибка. Не пытайтесь.

⚡ Предскажите **каждое** значение и впишите прогноз в ячейку, потом запускайте.


````python
# Мой прогноз: 

print(2 ** 3, 2 ** 3., 2. ** 3.)
print(6 / 3, 6 // 3, 6 // 3.)
print(6 // 4, -6 // 4, 6. // -4)
print(14 % 4, 12 % 4.5)
````

### Приоритеты и связывание

Умножение выполняется раньше сложения: в `2 + 3 * 5` получится `17`, а не `25`. Это называется **иерархией приоритетов**.

| Приоритет | Оператор | Тип |
|---|---|---|
| 1 | `**` | |
| 2 | `+`, `-` | унарные |
| 3 | `*`, `/`, `//`, `%` | |
| 4 | `+`, `-` | бинарные |

Если приоритет одинаковый, работает **связывание** (ассоциативность). У большинства операторов Python оно **левостороннее** (слева направо), но у `**` - **правостороннее**.

**Скобки** меняют порядок вычислений: подвыражение в скобках вычисляется первым. Скобок можно ставить сколько угодно, в том числе просто для читаемости.

⚡ Здесь легко ошибиться. Предскажите каждую строку и впишите прогноз в ячейку.


````python
# Мой прогноз: 

print(9 % 6 % 2)
print(2 ** 2 ** 3)
print(-3 ** 2)
print(-(3 ** 2))
print(2 * 3 % 5)
````

## Часть 3. Переменные (2.4)

**Переменная** - именованная «коробка» для значения. У неё есть **имя** и **значение**.

**Переменная возникает в момент присваивания ей значения** - объявлять её отдельно не нужно:

```python
имя = значение
```

**Правила именования:**

- только буквы, цифры и `_` (подчёркивание считается буквой);
- имя начинается с буквы (или с `_`), но не с цифры;
- регистр важен: `var` и `Var` - две **разные** переменные;
- имя не должно совпадать с ключевым словом Python;
- пробелов в имени быть не может.

Корректно: `counter`, `_`, `t34`, `days_to_christmas`, `Exchange_Rate`, `переменная`.
Некорректно: `10t`, `!important`, `exchange rate`.

По стандарту [PEP 8](https://peps.python.org/pep-0008/) имена переменных и функций пишут строчными буквами, разделяя слова подчёркиванием: `my_variable`.

Ключевые слова Python можно не заучивать - их всегда можно спросить у самого Python. Запустите:

````python
import keyword
print(keyword.kwlist)
````

Знак `=` - это **оператор присваивания**, а не знак равенства. Он берёт значение справа и кладёт его в переменную слева. Поэтому запись `var = var + 1` не абсурдна: она означает «взять текущее значение `var`, прибавить `1`, результат снова положить в `var`».

⚡ Что выведут эти ячейки?

````python
# Мой прогноз: 

var = 1
print(var)
var = var + 1
print(var)
````

````python
# Мой прогноз: 

var = 100
var = 200 + 300
print(var)
````

### Сокращённые операторы

Если переменная стоит и слева, и справа от `=`, запись можно сократить. Общее правило: `variable = variable op expression` записывается как `variable op= expression`.

| Полная запись | Сокращённая |
|---|---|
| `i = i + 2 * j` | `i += 2 * j` |
| `var = var / 2` | `var /= 2` |
| `rem = rem % 10` | `rem %= 10` |
| `x = x ** 2` | `x **= 2` |

Решают проблему три функции преобразования типов:

`int()` и `float()` завершат программу с ошибкой, если преобразовать не удалось. `str()` безопасна: число в строку превращается всегда.

### Строковые операторы

У `+` и `*` есть вторая профессия - работа со строками.

**`+` - конкатенация** (склеивание). Оба аргумента должны быть строками, смешивать типы нельзя. Оператор **не коммутативен**: `"ab" + "ba"` не то же самое, что `"ba" + "ab"`.

**`*` - репликация** (повторение). Работает со строкой и числом в любом порядке:

- `"Гоша" * 3` даёт `"ГошаГошаГоша"`
- `5 * "2"` даёт `"22222"` (не `10`!)
- число меньше или равно нулю даёт **пустую строку**.

⚡ Что нарисует эта программа? Обратите внимание на скобки во второй строке.

````python
# Мой прогноз: 

print("+" + 10 * "-" + "+")
print(("|" + " " * 10 + "|\n") * 5, end="")
print("+" + 10 * "-" + "+")
````
