## 3.2.1 Циклы с while

Согласны ли вы со следующим утверждением?

```
пока есть что делать
    делай это
```

Обратите внимание: эта запись также означает, что если делать нечего, не произойдёт вообще ничего.

В общем виде цикл в Python можно представить так:

```
while:
    инструкция
```

Если вы заметили некоторое сходство с инструкцией `if`, то вы совершенно правы. Действительно, синтаксическое отличие лишь одно: вместо слова `if` используется слово `while`.

Семантическое различие важнее: когда условие выполнено, `if` выполняет свои инструкции **только один раз**; `while` **повторяет выполнение** до тех пор, пока условие принимает значение `True`.

Обратите внимание: все правила про **отступы** действуют и здесь. Мы скоро это покажем.

Взгляните на алгоритм ниже:

```
while условное_выражение:
    инструкция_один
    инструкция_два
    инструкция_три
    :
    :
    инструкция_n
```

Теперь важно запомнить, что:

- если внутри одного цикла `while` нужно выполнить больше одной инструкции, все инструкции в блоке нужно (как и в случае с `if`) **оформить с одинаковым отступом**;
- инструкция или набор инструкций, выполняемых внутри цикла `while`, называется **телом цикла**;
- если условие - `False` (равно нулю) уже при первой проверке, тело не выполнится ни разу (обратите внимание на аналогию: если делать нечего, ничего и не нужно делать);
- тело должно быть способно изменить значение условия, потому что если условие изначально `True`, тело может выполняться бесконечно - обратите внимание, что обычно выполнение какого-то действия уменьшает количество оставшихся дел.


## 3.2.2 Бесконечный цикл

Бесконечный цикл, также называемый **вечным циклом**, - это последовательность инструкций в программе, которая повторяется бесконечно (зацикливается навсегда).

Вот пример цикла, который не может завершить своё выполнение:


````python
while True:
    print("Я застрял внутри цикла.")
````

Этот цикл будет бесконечно выводить на экран строку `"Я застрял внутри цикла."`.


Вернёмся к наброску алгоритма, который мы недавно показывали. Сейчас мы покажем, как использовать только что изученный цикл, чтобы найти наибольшее число среди большого набора введённых данных.

Внимательно проанализируйте программу. Обратите внимание, где начинается цикл (строка 8). Найдите тело цикла и разберитесь, **как происходит выход из тела цикла**:


````python
# Здесь храним текущее наибольшее число.
largest_number = -999999999

# Вводим первое значение.
number = int(input("Введите число или введите -1, чтобы остановиться: "))

# Если число не равно -1, продолжаем.
while number != -1:
    # Больше ли number, чем largest_number?
    if number > largest_number:
        # Да, обновляем largest_number.
        largest_number = number
    # Вводим следующее число.
    number = int(input("Введите число или введите -1, чтобы остановиться: "))

# Выводим наибольшее число.
print("Наибольшее число:", largest_number)
````

Проверьте, как этот код реализует алгоритм, который мы показывали раньше.


## 3.2.3 Цикл while: ещё примеры

Рассмотрим ещё один пример с использованием цикла `while`. Следите за комментариями, чтобы понять идею и решение.


````python
# Программа считывает последовательность чисел
# и считает, сколько из них чётных, а сколько - нечётных.
# Программа завершается, когда введён ноль.

odd_numbers = 0
even_numbers = 0

# Считываем первое число.
number = int(input("Введите число или введите 0, чтобы остановиться: "))

# 0 завершает выполнение.
while number != 0:
    # Проверяем, нечётное ли число.
    if number % 2 == 1:
        # Увеличиваем счётчик нечётных чисел.
        odd_numbers += 1
    else:
        # Увеличиваем счётчик чётных чисел.
        even_numbers += 1
    # Считываем следующее число.
    number = int(input("Введите число или введите 0, чтобы остановиться: "))

# Выводим результаты.
print("Количество нечётных чисел:", odd_numbers)
print("Количество чётных чисел:", even_numbers)
````

Некоторые выражения можно упростить, не меняя поведения программы.

Попробуйте вспомнить, как Python трактует истинность условия, и обратите внимание, что эти две формы эквивалентны:

`while number != 0:` и `while number:`.

Условие, проверяющее, является ли число нечётным, тоже можно записать в этих эквивалентных формах:

`if number % 2 == 1:` и `if number % 2:`.

### Использование переменной-счётчика для выхода из цикла

Взгляните на фрагмент ниже:


````python
counter = 5
while counter != 0:
    print("Внутри цикла.", counter)
    counter -= 1
print("Вне цикла.", counter)
````

Этот код должен вывести строку `"Внутри цикла."` и значение, хранящееся в переменной `counter`, во время выполнения цикла ровно пять раз. Как только условие перестаёт выполняться (переменная `counter` достигла `0`), цикл завершается, и выводится сообщение `"Вне цикла."`, а также значение, сохранённое в `counter`.

Но кое-что можно записать компактнее - условие цикла `while`.

Видите разницу?


````python
counter = 5
while counter:
    print("Внутри цикла.", counter)
    counter -= 1
print("Вне цикла.", counter)
````

Стало ли это компактнее, чем раньше? Немного. Стало ли понятнее? Тут можно поспорить.

## 3.2.5 Циклы с for

Ещё один вид цикла, доступный в Python, происходит из наблюдения, что иногда важнее **считать «обороты» цикла**, чем проверять условия.

Представьте, что тело цикла должно быть выполнено ровно сто раз. Если бы вы захотели сделать это с помощью цикла `while`, это могло бы выглядеть так:


````python
i = 0
while i < 100:
    # do_something()
    i += 1
````

Было бы неплохо, если бы кто-то мог сделать это скучное считание за вас. Возможно ли это?

Конечно, возможно - для таких задач есть особый цикл, и называется он `for`.

На самом деле цикл `for` предназначен для более сложных задач - **он умеет «просматривать» большие коллекции данных поэлементно**. Мы скоро покажем, как это делать, а пока рассмотрим более простой вариант его применения.

Взгляните на фрагмент:


````python
for i in range(100):
    # do_something()
    pass
````

Здесь есть несколько новых элементов. Давайте о них расскажем:

- ключевое слово `for` открывает цикл `for`; обратите внимание - после него нет условия; вам не нужно думать об условиях, поскольку они проверяются внутренне, автоматически;
- любая переменная после ключевого слова `for` - это **управляющая переменная** цикла; она автоматически считает «обороты» цикла;
- ключевое слово `in` вводит синтаксический элемент, описывающий диапазон возможных значений, присваиваемых управляющей переменной;
- функция `range()` (это очень особенная функция) отвечает за генерацию всех нужных значений управляющей переменной; в нашем примере функция будет создавать (можно даже сказать - «скармливать» циклу) последовательные значения из следующего набора: 0, 1, 2 .. 97, 98, 99; обратите внимание: в этом случае функция `range()` начинает работу с 0 и заканчивает на один шаг (одно целое число) раньше значения своего аргумента;
- обратите внимание на ключевое слово `pass` внутри тела цикла - оно вообще ничего не делает; это **пустая инструкция** - мы поставили её здесь, потому что синтаксис цикла `for` требует хотя бы одной инструкции внутри тела (кстати, то же самое верно и для `if`, `elif`, `else` и `while`).

Наши следующие примеры будут скромнее по количеству повторений цикла.

Взгляните на фрагмент ниже. Сможете предсказать, что он выведет?


````python
for i in range(10):
    print("Значение i сейчас равно", i)
````

Запустите код и проверьте, правы ли вы.

Обратите внимание:

- цикл выполнился десять раз (это аргумент функции `range()`)
- последнее значение управляющей переменной - `9` (не `10`, поскольку отсчёт начинается **с 0**, а не с 1).

Вызов функции `range()` можно снабдить двумя аргументами, а не одним:


````python
for i in range(2, 8):
    print("Значение i сейчас равно", i)
````

В этом случае первый аргумент определяет начальное (первое) значение управляющей переменной.

Последний аргумент показывает первое значение, которое **не будет** присвоено управляющей переменной.

Обратите внимание: функция `range()` **принимает в качестве аргументов только целые числа** и генерирует последовательности целых чисел.

## 3.2.6 Ещё о цикле for и функции range() с тремя аргументами

Функция `range()` может принимать и **три аргумента** - взгляните на код в редакторе.


````python
for i in range(2, 8, 3):
    print("Значение i сейчас равно", i)
````

Третий аргумент - это **приращение** (шаг) - значение, добавляемое к управляющей переменной на каждом обороте цикла (как вы, наверное, догадываетесь, **значение приращения по умолчанию равно 1**).

Сможете сказать, сколько строк появится в консоли и что они будут содержать?

Запустите программу, чтобы проверить, правы ли вы.

Вы должны увидеть в окне консоли следующие строки:

```
Значение i сейчас равно 2
Значение i сейчас равно 5
```

Знаете, почему? Первый аргумент, переданный функции `range()`, сообщает нам, каково **начальное** число последовательности (отсюда `2` в выводе). Второй аргумент говорит функции, где **остановить** последовательность (функция генерирует числа до числа, указанного вторым аргументом, но не включая его). Наконец, третий аргумент указывает **шаг**, который на самом деле означает разницу между каждым числом в последовательности, создаваемой функцией.

`2` (начальное число) → `5` (`2` увеличено на `3` равно `5` - число входит в диапазон от 2 до 8) → `8` (`5` увеличено на `3` равно `8` - это число уже не входит в диапазон от 2 до 8, поскольку стоп-параметр не включается в последовательность чисел, создаваемую функцией.)

Обратите внимание: если набор, создаваемый функцией `range()`, пуст, цикл вообще не выполнит своё тело.

Вот так - здесь не будет никакого вывода:


````python
for i in range(1, 1):
    print("Значение i сейчас равно", i)
````

Обратите внимание: набор, создаваемый функцией `range()`, должен быть отсортирован **по возрастанию**. Нет способа заставить `range()` создать набор в другой форме, если функции `range()` передаются ровно два аргумента. Это значит, что второй аргумент `range()` должен быть больше первого.

Поэтому здесь тоже не будет никакого вывода:


````python
for i in range(2, 1):
    print("Значение i сейчас равно", i)
````

````python
power = 1
for expo in range(16):
    print("2 в степени", expo, "равно", power)
    power *= 2
````

## 3.2.8 Инструкции break и continue

До сих пор мы рассматривали тело цикла как неделимую и неразрывную последовательность инструкций, которая полностью выполняется на каждом обороте цикла. Однако, как разработчик, вы можете столкнуться с такими ситуациями:

- оказывается, что цикл в целом больше не нужно продолжать; следует прекратить дальнейшее выполнение тела цикла и пойти дальше;
- оказывается, что нужно начать следующий оборот цикла, не завершая выполнение текущего оборота.

Python предоставляет две особые инструкции для реализации обеих этих задач. Скажем для точности, что их наличие в языке не является необходимым - опытный программист способен закодировать любой алгоритм и без этих инструкций. Такие дополнения, которые не расширяют выразительную мощь языка, а лишь упрощают работу разработчика, иногда называют **синтаксическим сахаром**.

Вот эти две инструкции:

- `break` - немедленно выходит из цикла и безусловно завершает его работу; программа начинает выполнять ближайшую инструкцию после тела цикла;
- `continue` - ведёт себя так, как будто программа внезапно достигла конца тела цикла; начинается следующий оборот, и сразу же проверяется условное выражение.

Оба этих слова - **ключевые слова**.

Сейчас мы покажем два простых примера, иллюстрирующих работу этих двух инструкций. Взгляните на код в редакторе. Запустите программу и проанализируйте вывод. Измените код и поэкспериментируйте.


````python
# break - пример

print("Инструкция break:")
for i in range(1, 6):
    if i == 3:
        break
    print("Внутри цикла.", i)
print("Вне цикла.")

# continue - пример

print("\nИнструкция continue:")
for i in range(1, 6):
    if i == 3:
        continue
    print("Внутри цикла.", i)
print("Вне цикла.")
````

> *Примечание: код примера с `break` реконструирован по аналогии с примером `continue` ниже него - в редакторе была видна только нижняя, прокрученная часть виджета.*

### Инструкции break и continue: ещё примеры

Вернёмся к нашей программе, которая определяет наибольшее из введённых чисел. Мы переделаем её дважды, используя инструкции `break` и `continue`.

Проанализируйте код и подумайте, стали бы вы и как использовать любую из них.

Вариант с `break` выглядит так:


````python
counter = 0
largest_number = -999999999
while True:
    number = int(input("Введите число или введите -1, чтобы завершить программу: "))
    if number == -1:
        break
    counter += 1
    if number > largest_number:
        largest_number = number

if counter != 0:
    print("Наибольшее число:", largest_number)
else:
    print("Вы не ввели ни одного числа.")
````

Запустите его, протестируйте и поэкспериментируйте с ним.

А теперь вариант с `continue`:


````python
counter = 0
largest_number = -999999999
number = int(input("Введите число или введите -1, чтобы завершить программу: "))

while True:
    if number == -1:
        continue
    counter += 1

    if number > largest_number:
        largest_number = number
    number = int(input("Введите число или введите -1, чтобы завершить программу: "))

    if counter:
        print("Наибольшее число:", largest_number)
    else:
        print("Вы не ввели ни одного числа.")
````

> *Примечание: в этом фрагменте, судя по описанию оригинала, пользователь вводит первое число ещё **до** того, как программа войдёт в цикл `while`, а последующее число вводится, когда программа уже находится внутри цикла - структура кода выше реконструирована по этому описанию, поскольку виджет редактора в тексте урока был прокручен и часть кода не попала в выгрузку.*
