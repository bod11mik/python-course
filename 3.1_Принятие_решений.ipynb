## 3.1.2 Сравнение: оператор равенства

Вопрос: **равны ли два значения**?

Чтобы задать этот вопрос, используется оператор `==` (равно-равно).

Не забывайте про важное различие:

- `=` - это **оператор присваивания**, например, `a = b` присваивает `a` значение `b`;
- `==` - это вопрос *равны ли эти значения?*, то есть `a == b` **сравнивает** `a` и `b`.

Это бинарный оператор с левосторонним связыванием. Он принимает два аргумента и **проверяет, равны ли они**.

## 3.1.4 Операторы

### Равенство: оператор *равно* (==)

Оператор `==` (равно) сравнивает значения двух операндов. Если они равны, результат сравнения - `True`. Если не равны, результат - `False`.

Взгляните на сравнение на равенство ниже - каким будет результат этой операции?

```
var == 0
```

Обратите внимание: мы не можем найти ответ, если не знаем, какое значение сейчас хранится в переменной `var`.

Если переменная менялась много раз за время выполнения программы, или её начальное значение вводится с консоли, ответить на этот вопрос может только сам Python, и только во время выполнения программы.

Теперь представьте программиста, который страдает от бессонницы и вынужден считать чёрных и белых овец по отдельности, пока чёрных овец ровно вдвое больше, чем белых.

Вопрос будет звучать так:

```
black_sheep == 2 * white_sheep
```

Из-за низкого приоритета оператора `==` этот вопрос будет трактоваться как эквивалентный следующему:

```
black_sheep == (2 * white_sheep)
```

Итак, давайте попрактикуемся в понимании оператора `==` - сможете угадать, что выведет код ниже?


````python
var = 0  # Присваиваем var значение 0
print(var == 0)

var = 1  # Присваиваем var значение 1
print(var == 0)
````

### Неравенство: оператор *не равно* (!=)

Оператор `!=` (не равно) тоже сравнивает значения двух операндов. Разница вот в чём: если они равны, результат сравнения - `False`. Если не равны - результат `True`.

Теперь взгляните на сравнение на неравенство ниже - сможете угадать результат этой операции?


````python
var = 0  # Присваиваем var значение 0
print(var != 0)

var = 1  # Присваиваем var значение 1
print(var != 0)
````

У оператора *больше* есть особый, **нестрогий** вариант, но обозначается он иначе, чем в классической математической записи: `>=` (больше или равно).

Здесь два символа подряд, а не один.

Оба этих оператора (строгий и нестрогий), как и два других, о которых пойдёт речь дальше, - **бинарные операторы с левосторонним связыванием**, и их приоритет выше, чем у `==` и `!=`.

Если мы хотим узнать, нужно ли нам надевать тёплую шапку, зададим такой вопрос:

Как вы, наверное, уже догадались, для этого случая используются операторы `<` (меньше) и его нестрогий «родственник»: `<=` (меньше или равно).

Взгляните на этот простой пример:


Мы проверяем, есть ли риск получить штраф от дорожной полиции (первый вопрос - строгий, второй - нет).

## 3.1.7 Условия и условное выполнение

Вы уже знаете, как задавать Python вопросы, но пока не знаете, как разумно использовать полученные ответы. Вам нужен механизм, который позволит **сделать что-то, если условие выполнено, и не делать этого, если нет**.

Точно так же и в жизни: вы делаете что-то или не делаете, в зависимости от того, выполняется ли определённое условие, например, вы идёте гулять, если погода хорошая, или остаётесь дома, если на улице сыро и холодно.

Чтобы принимать такие решения, Python предлагает особую инструкцию. Из-за своей природы и применения она называется **условной инструкцией** (или условным оператором).

У неё есть несколько вариантов. Начнём с самого простого, постепенно усложняя.

Первая форма условного оператора, которую вы видите ниже, записана очень неформально, но наглядно:

```
if true_or_not:
    do_this_if_true
```

Этот условный оператор состоит из следующих, строго обязательных элементов, именно в таком порядке:

- ключевое слово `if`;
- один или несколько пробелов;
- выражение (вопрос или ответ), значение которого будет трактоваться исключительно в терминах `True` (когда оно ненулевое) и `False` (когда оно равно нулю);
- **двоеточие**, за которым следует переход на новую строку;
- инструкция или набор инструкций **с отступом** (обязательна хотя бы одна инструкция); **отступ** можно сделать двумя способами - вставив определённое количество пробелов (рекомендуется использовать **отступ в четыре пробела**), или используя символ табуляции (*tab*); обратите внимание: если в блоке с отступом больше одной инструкции, отступ должен быть **абсолютно одинаковым** во всех строках; даже если визуально это выглядит одинаково при смешивании пробелов и табуляции, важно, чтобы все отступы были **совершенно одинаковыми** - Python 3 **не допускает смешивания пробелов и табуляции** для отступов.

Как работает эта инструкция?

- если выражение `true_or_not` **представляет истину** (то есть его значение не равно нулю), инструкция(и) с отступом **будет выполнена**;
- если выражение `true_or_not` **не представляет истину** (то есть его значение равно нулю), инструкция(и) с отступом **будет пропущена** (проигнорирована), и следующей выполненной инструкцией будет та, что стоит после исходного уровня отступа.

В жизни мы часто выражаем желание так:

*если погода хорошая, мы пойдём гулять*

*потом мы пообедаем*

Как видите, обед - это не условное действие, и от погоды он не зависит.

Зная, какие условия влияют на наше поведение, и предположив, что у нас есть безаргументные функции `go_for_a_walk()` и `have_lunch()`, можно написать такой фрагмент:


````python
if the_weather_is_good:
    go_for_a_walk()
have_lunch()
````

### Условное выполнение: инструкция if

Если некий страдающий бессонницей разработчик на Python засыпает, когда досчитает до 120 овец, а процедуру, вызывающую сон, можно реализовать в виде особой функции с именем `sleep_and_dream()`, весь код будет выглядеть так:


````python
if sheep_counter >= 120: # Вычисляем проверочное выражение
    sleep_and_dream() # Выполняем, если проверочное выражение истинно
````

Это можно прочитать так: если `sheep_counter` больше или равно `120`, то засыпаем и видим сон (то есть выполняем функцию `sleep_and_dream`).

Мы уже говорили, что **условно выполняемые инструкции должны иметь отступ**. Это создаёт очень наглядную структуру, ясно показывающую все возможные пути выполнения кода.

Взгляните на следующий код:


````python
if sheep_counter >= 120:
    make_a_bed()
    take_a_shower()
    sleep_and_dream()
feed_the_sheepdogs()
````

Как видите, застилание постели, принятие душа, засыпание и сон - всё это выполняется условно - когда `sheep_counter` достигает нужного предела.

А вот кормление овчарок выполняется всегда (то есть функция `feed_the_sheepdogs()` не имеет отступа и не принадлежит блоку `if`, а значит, выполняется всегда).

Теперь обсудим другой вариант условного оператора, который также позволяет выполнить дополнительное действие, когда условие не выполнено.

### Условное выполнение: инструкция if-else

Мы начали с простой фразы: *Если погода хорошая, мы пойдём гулять.*

Обратите внимание: тут ничего не сказано о том, что будет, если погода плохая. Мы знаем только, что не пойдём на улицу, но что мы будем делать вместо этого - неизвестно. Возможно, нам захочется запланировать что-то и на случай плохой погоды.

Можно сказать, например: *Если погода хорошая, мы пойдём гулять, иначе пойдём в кино.*

Теперь мы знаем, что будем делать, **если условия выполнены**, и знаем, что будем делать, **если что-то пошло не так**. Другими словами, у нас есть «план Б».

Python позволяет выразить такие альтернативные планы. Для этого используется вторая, чуть более сложная форма условного оператора - инструкция *if-else*:


````python
if true_or_false_condition:
    perform_if_condition_true
else:
    perform_if_condition_false
````

Итак, появляется новое слово: `else` - это ключевое слово.

Часть кода, которая начинается с `else`, описывает, что делать, если условие, заданное для `if`, не выполнено (обратите внимание на двоеточие после этого слова).

Выполнение *if-else* происходит так:

- если условие принимает значение `True` (его значение не равно нулю), выполняется инструкция `perform_if_condition_true`, и на этом условный оператор завершается;
- если условие принимает значение `False` (оно равно нулю), выполняется инструкция `perform_if_condition_false`, и на этом условный оператор завершается.

### Инструкция if-else: ещё немного условного выполнения

Используя эту форму условного оператора, можно описать наши планы так:


````python
if the_weather_is_good:
    go_for_a_walk()
else:
    go_to_a_theater()
have_lunch()
````

Если погода хорошая, мы пойдём гулять. Иначе - пойдём в кино. В любом случае, независимо от того, хорошая погода или плохая, мы потом пообедаем (после прогулки или после похода в кино).

Всё, что мы говорили про отступы, действует точно так же и внутри ветки `else`:


````python
if the_weather_is_good:
    go_for_a_walk()
    have_fun()
else:
    go_to_a_theater()
    enjoy_the_movie()
have_lunch()
````

### Вложенные инструкции if-else

Теперь обсудим два особых случая условного оператора.

Во-первых, рассмотрим случай, когда **инструкция, стоящая после if, - это ещё один if**.

Прочитайте, что мы запланировали на это воскресенье. Если погода хорошая, мы пойдём гулять. Если найдём хороший ресторан, пообедаем там. Иначе перекусим бутербродом. Если погода плохая, пойдём в кино. Если билетов не будет, поедем за покупками в ближайший торговый центр.

Запишем то же самое на Python. Внимательно рассмотрите этот код:


````python
if the_weather_is_good:
    if nice_restaurant_is_found:
        have_lunch()
    else:
        eat_a_sandwich()
else:
    if tickets_are_available:
        go_to_the_theater()
    else:
        go_shopping()
````

Здесь важны два момента:

- такое использование `if` называется **вложенностью** (nesting); помните, что каждый `else` относится к тому `if`, который находится **на том же уровне отступа** - это нужно знать, чтобы понять, как `if` и `else` соотносятся друг с другом;
- обратите внимание, как **отступ улучшает читаемость** и делает код понятнее для отслеживания.

### Инструкция elif

Второй особый случай знакомит нас с ещё одним новым ключевым словом Python: `elif`. Как вы уже, наверное, догадались, это сокращённая форма `else if`.

`elif` используется, чтобы **проверить больше одного условия**, и **останавливается**, как только находит первую истинную инструкцию.

Наш следующий пример похож на вложенность, но сходство лишь незначительное. Снова изменим наши планы и опишем их так: если погода хорошая, мы пойдём гулять, иначе, если достанем билеты, пойдём в кино, иначе, если в ресторане будут свободные столики, пойдём на обед; если же ничего не получится, останемся дома и сыграем в шахматы.

Заметили, сколько раз мы использовали слово *иначе*? Именно здесь и проявляется роль ключевого слова `elif`.

Запишем этот же сценарий на Python:


````python
if the_weather_is_good:
    go_for_a_walk()
elif tickets_are_available:
    go_to_the_theater()
elif table_is_available:
    go_for_lunch()
else:
    play_chess_at_home()
````

Способ сборки последовательных инструкций *if-elif-else* иногда называют **каскадом**.

Обратите внимание, как отступ снова улучшает читаемость кода.

Стоит уделить внимание и следующим моментам:

- **нельзя использовать `else` без предшествующего `if`**;
- `else` всегда является **последней веткой каскада**, независимо от того, использовался `elif` или нет;
- `else` - это **необязательная** часть каскада, её можно опустить;
- если в каскаде есть ветка `else`, выполняется **только одна** из всех веток;
- если ветки `else` нет, возможна ситуация, что **не будет выполнена ни одна** из доступных веток.

Это может звучать немного запутанно, но, надеемся, несколько простых примеров прояснят ситуацию.


## 3.1.8 Анализ примеров кода

Сейчас мы покажем несколько простых, но полноценных программ. Мы не будем разбирать их подробно, поскольку считаем, что комментариев (и имён переменных) внутри кода вполне достаточно, чтобы всё было понятно.

Все программы решают одну и ту же задачу - они **находят наибольшее из нескольких чисел и выводят его**.

**Пример 1:**

Начнём с самого простого случая - **как определить, какое из двух чисел больше**:


````python
# Считываем два числа
number1 = int(input("Введите первое число: "))
number2 = int(input("Введите второе число: "))

# Выбираем большее число
if number1 > number2:
    larger_number = number1
else:
    larger_number = number2

# Выводим результат
print("Большее число:", larger_number)
````

Приведённый выше фрагмент должен быть понятен - он считывает два целочисленных значения, сравнивает их и находит большее.

**Пример 2:**

Теперь покажем вам один любопытный факт. У Python есть интересная особенность - взгляните на код ниже:


````python
# Считываем два числа
number1 = int(input("Введите первое число: "))
number2 = int(input("Введите второе число: "))

# Выбираем большее число
if number1 > number2: larger_number = number1
else: larger_number = number2

# Выводим результат
print("Большее число:", larger_number)
````

Обратите внимание: если какая-то из веток *if-elif-else* содержит всего одну инструкцию, её можно записать в более компактной форме (не нужно делать отдельную строку с отступом после ключевого слова - можно просто продолжить строку после двоеточия).

Впрочем, такой стиль может сбивать с толку, и в наших дальнейших программах мы не будем его использовать. Но знать о нём определённо стоит, если вы хотите читать и понимать чужие программы.

Больше никаких отличий в коде нет.

**Пример 3:**

Пришло время усложнить код - найдём наибольшее из трёх чисел. Увеличится ли код? Немного.

Предположим, что наибольшее - первое значение. Затем проверим эту гипотезу на двух оставшихся значениях.

Взгляните на код ниже:


````python
# Считываем три числа
number1 = int(input("Введите первое число: "))
number2 = int(input("Введите второе число: "))
number3 = int(input("Введите третье число: "))

# Временно предполагаем, что первое число -
# наибольшее.
# Скоро мы это проверим.
largest_number = number1

# Проверяем, больше ли второе число текущего largest_number,
# и при необходимости обновляем largest_number.
if number2 > largest_number:
    largest_number = number2

# Проверяем, больше ли третье число текущего largest_number,
# и при необходимости обновляем largest_number.
if number3 > largest_number:
    largest_number = number3

# Выводим результат
print("Наибольшее число:", largest_number)
````

Этот способ значительно проще, чем пытаться найти наибольшее число сразу, сравнивая все возможные пары чисел (то есть первое со вторым, второе с третьим, третье с первым). Попробуйте воссоздать этот код самостоятельно.


## 3.1.9 Псевдокод и введение в циклы

Теперь вы уже должны уметь написать программу, которая находит наибольшее из четырёх, пяти, шести или даже десяти чисел.

Схему вы уже знаете, так что расширить масштаб задачи будет несложно.

Но что если мы попросим вас написать программу, которая находит наибольшее из двухсот чисел? Можете представить такой код?

Вам понадобится двести переменных. А если двухсот переменных мало, попробуйте представить поиск наибольшего среди миллиона чисел.

Представьте код, содержащий 199 условных операторов и двести вызовов функции `input()`. К счастью, вам не придётся с этим разбираться. Есть способ проще.

Пока забудем о требованиях синтаксиса Python и попробуем проанализировать задачу, не думая о самом программировании. Другими словами, попробуем записать **алгоритм**, и когда он нас устроит, реализуем его.

В этом случае мы воспользуемся особой записью, которая не является настоящим языком программирования (её нельзя ни скомпилировать, ни выполнить), но она формализована, лаконична и легко читается. Она называется **псевдокодом**.

Посмотрите на наш псевдокод ниже:

```
largest_number = -999999999
number = int(input())
if number == -1:
    print(largest_number)
    exit()
if number > largest_number:
    largest_number = number
# Перейти к строке 02
```

Что здесь происходит?

Во-первых, можно упростить программу, если в самом начале кода присвоить переменной `largest_number` значение, которое будет меньше любого из вводимых чисел. Для этого мы используем `-999999999`.

Во-вторых, мы предполагаем, что наш алгоритм заранее не знает, сколько чисел будет введено. Мы ожидаем, что пользователь введёт столько чисел, сколько захочет - алгоритм будет одинаково хорошо работать и со ста, и с тысячей чисел. Как это сделать?

Мы договоримся с пользователем: когда будет введено значение `-1`, это будет знаком, что данных больше нет, и программа должна завершить работу.

В противном случае, если введённое значение не равно `-1`, программа считает ещё одно число, и так далее.

Этот приём основан на предположении, что любую часть кода можно выполнить более одного раза - ровно столько раз, сколько нужно.

Выполнение определённой части кода более одного раза называется **циклом**. Смысл этого термина вам, вероятно, уже понятен.

Строки `02`-`08` образуют цикл. Мы будем **проходить их столько раз, сколько потребуется**, чтобы просмотреть все введённые значения.

Можно ли использовать похожую конструкцию в программе на Python? Да, можно.

> **Дополнительная информация.** В Python часто есть много встроенных функций, которые сделают эту работу за вас. Например, чтобы найти наибольшее число из всех, можно использовать встроенную функцию Python под названием `max()`. Её можно использовать с несколькими аргументами. Проанализируйте код ниже:


````python
# Считываем три числа.
number1 = int(input("Введите первое число: "))
number2 = int(input("Введите второе число: "))
number3 = int(input("Введите третье число: "))

# Проверяем, какое из чисел наибольшее,
# и передаём его в переменную largest_number.

largest_number = max(number1, number2, number3)

# Выводим результат.
print("Наибольшее число:", largest_number)
````

Точно так же можно использовать функцию `min()`, чтобы вернуть наименьшее число. Можете переделать код выше и поэкспериментировать с ним в песочнице (Sandbox).
