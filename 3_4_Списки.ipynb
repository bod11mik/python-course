# 3.4 Списки

Списки Python: индексация, изменение и удаление элементов, срезы, основные функции и методы.

## Зачем нужны списки?

Хранить сотни значений в отдельных переменных неудобно. Список - это одна переменная, которая хранит сразу несколько значений, пронумерованных начиная с 0.

Создаём список:

````python
numbers = [10, 5, 7, 2, 1]
````

Список - это коллекция, но каждый элемент внутри - обычный скаляр (число, строка и т.д.). Элементы могут быть разных типов.

## Индексация списков

Изменить значение элемента можно присвоением по индексу в квадратных скобках:

````python
numbers = [10, 5, 7, 2, 1]
print("Исходное содержимое списка:", numbers)

numbers[0] = 111
print("Новое содержимое списка:", numbers)
````

Скопируйте значение пятого элемента во второй - как это сделать?

````python
numbers[1] = numbers[4]  # Копируем значение пятого элемента во второй.
print("Новое содержимое списка:", numbers)
````

Индексом может быть не только число-литерал, но и любое выражение.

## Доступ к содержимому списка

К элементу можно обратиться по индексу:

````python
print(numbers[0])  # Обращение к первому элементу списка.
````

Список можно вывести и целиком:

````python
print(numbers)  # Вывод списка целиком.
````

Длина списка может меняться во время выполнения программы. Узнать текущую длину - функция `len()`. Угадайте вывод:

````python
print("Длина списка:", len(numbers))
````

## Удаление элементов: del

`del` - инструкция (не функция), которая удаляет элемент списка; длина списка уменьшается на 1. Угадайте вывод:

````python
del numbers[1]  # Удаляем второй элемент из списка.
print(len(numbers))
print(numbers)
````

Обратиться к удалённому индексу нельзя - ни прочитать, ни присвоить значение. Запустите код ниже и убедитесь, что он вызовет ошибку:

````python
print(numbers[4])
numbers[4] = 1
````

## Отрицательные индексы

Отрицательные индексы допустимы: `-1` - последний элемент списка.

````python
numbers = [111, 7, 2, 1]
print(numbers[-1])
````

А это?

````python
print(numbers[-2])
````

Последний доступный (в обратную сторону) индекс - `numbers[-len(numbers)]`, он указывает на первый элемент.

## Функции vs методы

Метод похож на функцию, но принадлежит конкретным данным и может менять их состояние:

```python
result = function(arg)      # функция
```

## Добавление элементов: append() и insert()

- `list.append(value)` - добавляет элемент в конец списка;
- `list.insert(location, value)` - вставляет элемент на нужную позицию, сдвигая остальные вправо.

Посмотрите на append() и insert() в действии:

````python
numbers = [111, 7, 2, 1]
print(len(numbers))
print(numbers)
````

````python
numbers.append(4)
print(len(numbers))
print(numbers)
````

````python
numbers.insert(0, 222)
print(len(numbers))
print(numbers)
````

Добавьте следующий фрагмент после последней строки кода:

````python
numbers.insert(1, 333)
print(len(numbers))
print(numbers)
````

Список можно начать с пустого (`[]`) и заполнять по мере необходимости. Угадайте вывод:

````python
my_list = []  # Создаём пустой список.

for i in range(5):
    my_list.append(i + 1)

print(my_list)
````

Немного изменим фрагмент - вместо append() используем insert(0, ...):

````python
my_list = []  # Создаём пустой список.

for i in range(5):
    my_list.insert(0, i + 1)

print(my_list)
````

## Как пользоваться списками

Цикл `for` умеет эффективно обрабатывать списки:

````python
my_list = [10, 1, 8, 3, 5]
total = 0

for i in range(len(my_list)):
    total += my_list[i]

print(total)
````

Переменная `total` накапливает сумму элементов. Есть и более короткая форма `for` - без индексов:

````python
my_list = [10, 1, 8, 3, 5]
total = 0

for i in my_list:
    total += i

print(total)
````

Здесь `for i in my_list` сразу перебирает значения элементов - индексы и `len()` не нужны.

## Списки в деле: обмен значениями

Чтобы поменять местами значения двух переменных, нужна вспомогательная переменная:

````python
variable_1 = 1
variable_2 = 2

auxiliary = variable_1
variable_1 = variable_2
variable_2 = auxiliary

print(variable_1, variable_2)
````

Python предлагает более удобный способ:

````python
variable_1 = 1
variable_2 = 2

variable_1, variable_2 = variable_2, variable_1

print(variable_1, variable_2)
````

Тот же приём работает и для элементов списка:

````python
my_list = [10, 1, 8, 3, 5]

my_list[0], my_list[4] = my_list[4], my_list[0]
my_list[1], my_list[3] = my_list[3], my_list[1]

print(my_list)
````

Запустите и проверьте вывод. Получится ли так же просто для списка из 100 элементов?

````python
my_list = [10, 1, 8, 3, 5]
length = len(my_list)

for i in range(length // 2):
    my_list[i], my_list[length - i - 1] = my_list[length - i - 1], my_list[i]

print(my_list)
````

Идея: `length // 2` итераций, на каждой меняем местами `i`-й элемент (от начала) с элементом `length - i - 1` (от конца).
